````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# 1. Нормализация городов: убрать пробелы по краям, привести к виду "Almaty"
cities_clean = [c.strip().title() for c in cities]
print(cities_clean)

# 2. Строки -> числа
amounts_int = [int(a) for a in amounts]
print(amounts_int)

# 3. Общие показатели
n_orders = len(amounts_int)
total = sum(amounts_int)
avg_check = round(total / n_orders)
print(f"Заказов: {n_orders}")
print(f"Общая сумма: {total} ₸")
print(f"Средний чек: {avg_check} ₸")

# 4. Неоплаченные заказы
unpaid_count = 0
unpaid_sum = 0
for amount, is_paid in zip(amounts_int, paid):
    if not is_paid:
        unpaid_count += 1
        unpaid_sum += amount
print(f"Не оплачено: {unpaid_count} заказов на {unpaid_sum} ₸")

# 5. Таблица
print(f"{'город':<10} | {'сумма':>6} | статус")
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    status = "оплачен" if is_paid else "не оплачен"
    print(f"{city:<10} | {amount:>6} | {status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# 6. Выручка (только оплаченные) по Алматы и Астане
rev_almaty = 0
rev_astana = 0
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    if is_paid and city == "Almaty":
        rev_almaty += amount
    elif is_paid and city == "Astana":
        rev_astana += amount
print(f"Алматы: {rev_almaty} ₸, Астана: {rev_astana} ₸")

# 7. Самый крупный оплаченный заказ (индекс и город)
best_idx = None
for i in range(len(amounts_int)):
    if paid[i] and (best_idx is None or amounts_int[i] > amounts_int[best_idx]):
        best_idx = i
print(f"Индекс: {best_idx}, город: {cities_clean[best_idx]}, сумма: {amounts_int[best_idx]} ₸")

# 8. Доля заказов выходного дня
weekend = 0
for d in days:
    if d in ("сб", "вс"):
        weekend += 1
print(f"Доля выходных: {weekend / len(days) * 100:.1f}%")

# 9. Копилка
goal = 400_000

# а) фиксированно 15 000 в месяц
saved, months = 0, 0
while saved < goal:
    saved += 15_000
    months += 1
print(f"Фиксированно 15 000: {months} мес.")

# б) каждый месяц на 1 000 больше, чем в прошлый
saved, months, deposit = 0, 0, 15_000
while saved < goal:
    saved += deposit
    months += 1
    deposit += 1_000
print(f"С ростом на 1 000: {months} мес. (накоплено {saved} ₸)")
````

## [Челлендж]
10. Без словарей и библиотек постройте два списка `unique_cities` и `revenue_by_city` (оплаченная выручка по каждому городу), отсортированные по убыванию выручки.
11. Найдите **медиану** суммы заказа без `statistics` и `numpy`. Учтите чётное число элементов.
12. Выведите текстовую «гистограмму» выручки по дням: каждые 5 000 ₸ рисуются как `█`.
```
пн | ████  20490
вт | ...
```
13. *Подумать:* какие проблемы данных вы встретили (пробелы, регистр, числа-строки)? Опишите 2–3 предложениями в Markdown-ячейке. Именно это мы будем делать в Pandas на занятии 6.

````python
# 10. Выручка по городам без словарей, сортировка по убыванию
unique_cities = []
for c in cities_clean:
    if c not in unique_cities:
        unique_cities.append(c)

revenue_by_city = [0] * len(unique_cities)
for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
    if is_paid:
        revenue_by_city[unique_cities.index(city)] += amount

# сортировка выбором: меняем местами элементы в обоих списках синхронно
for i in range(len(revenue_by_city)):
    m = i
    for j in range(i + 1, len(revenue_by_city)):
        if revenue_by_city[j] > revenue_by_city[m]:
            m = j
    revenue_by_city[i], revenue_by_city[m] = revenue_by_city[m], revenue_by_city[i]
    unique_cities[i], unique_cities[m] = unique_cities[m], unique_cities[i]

for c, r in zip(unique_cities, revenue_by_city):
    print(f"{c:<10} {r}")

# 11. Медиана без statistics / numpy
s = sorted(amounts_int)
n = len(s)
if n % 2 == 1:
    median = s[n // 2]
else:
    median = (s[n // 2 - 1] + s[n // 2]) / 2
print("Медиана:", median)

# 12. Гистограмма выручки (оплаченные заказы) по дням, 1 блок = 5 000 ₸
day_order = ["пн", "вт", "ср", "чт", "пт", "сб", "вс"]
for d in day_order:
    rev = 0
    for day, amount, is_paid in zip(days, amounts_int, paid):
        if day == d and is_paid:
            rev += amount
    print(f"{d} | {'█' * (rev // 5000):<10} {rev}")
````

**Ответ на 13 (эталон).** В данных встретились три проблемы: (1) лишние пробелы в начале и конце строк (`" almaty"`, `"Almaty "`); (2) разный регистр (`"SHYMKENT"`, `"astana"`), из-за чего один город считался бы за несколько разных; (3) суммы хранились как строки, поэтому их нельзя было складывать и сравнивать как числа (при сравнении строк `"9900" > "56000"`). Перед любым анализом данные нужно очистить, и именно это мы делаем в Pandas на занятии 6 (`str.strip()`, `str.title()`, `astype(int)`).
